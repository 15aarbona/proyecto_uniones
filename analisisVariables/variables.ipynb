## FILAS DE TORNILLOS POR ZONA
Todas las zonas se definen con **las mismas caras de las alas** (antes se mezclaban cara exterior, cara interior y eje del ala):

| zona | condición (posición `y` de la fila, mm desde el borde superior) |
|---|---|
| sobre el ala | `y < e + d` (siempre 1 fila: la de la cota `e`, a distancia `d` del ala) |
| **alma** | entre las caras **interiores** de las alas: `e+d+tf ≤ y ≤ e+d+h−tf` |
| **cantonera** | bajo la cara exterior del ala inferior (`y > e+d+h`), **sin contar la última fila** |
| bajo el último ala | la última fila (siempre 1) |

Una fila dentro del espesor de un ala sería una colisión: se detecta y se avisa.

```python
import warnings

p_cols = [f"p{i}" for i in range(1, 13)]
df[p_cols] = (
    df[p_cols].fillna(0.0).astype(float)
)  # p ausente = fila inactiva (esto sí es legítimo)

# Validación: antes los nulos se rellenaban con 0 en silencio y eso corrompía la geometría.
req = ["e", "d", "g", "BEAM_d_mm", "BEAM_tf_mm", "BEAM_tw_mm"]
bad_req = df[req].isna().any(axis=1) | (df[req] <= 0).any(axis=1)
assert not bad_req.any(), (
    f"{int(bad_req.sum())} filas con e/d/g/dimensiones de viga ausentes o <= 0"
)
assert not (np.diff((df[p_cols] > 0).values.astype(int), axis=1) > 0).any(), (
    "filas con p activas no consecutivas"
)


def calcular_geometria(df, tol=0.5):
    """Posiciones de las filas (mm) y su zona. Todo en arrays alineados con las filas de df."""
    e = df["e"].values.astype(float)
    active = (df[p_cols] > 0).values
    exists = np.column_stack(
        [np.ones(len(df), bool), active]
    )  # fila 1 (cota e) + p activas
    y = np.column_stack([e, e[:, None] + df[p_cols].cumsum(axis=1).values])
    y = np.where(exists, y, np.nan)
    top_out = e + df["d"].values  # cara exterior del ala superior
    top_in = top_out + df["BEAM_tf_mm"].values  # cara interior del ala superior
    bot_out = top_out + df["BEAM_d_mm"].values  # cara exterior del ala inferior
    bot_in = bot_out - df["BEAM_tf_mm"].values  # cara interior del ala inferior
    col = lambda v: v[:, None]
    sobre = exists & (y < col(top_out) - tol)
    alma = exists & (y >= col(top_in) - tol) & (y <= col(bot_in) + tol)
    en_ala = exists & ~sobre & ~alma & (y <= col(bot_out) + tol)
    bajo = exists & (y > col(bot_out) + tol)
    y_ult = np.nanmax(y, axis=1)
    return dict(
        y=y,
        exists=exists,
        D=y_ult,
        top_out=top_out,
        top_in=top_in,
        bot_out=bot_out,
        bot_in=bot_in,
        sobre=sobre,
        alma=alma,
        en_ala=en_ala,
        bajo=bajo,
        y_ult=y_ult,
        n_filas=exists.sum(axis=1),
        n_alma=alma.sum(axis=1),
        n_cant=np.maximum(bajo.sum(axis=1) - 1, 0),
    )


geo = calcular_geometria(df)
has_Lh = (df["Lh"].fillna(0) > 0).values
checks = pd.DataFrame(
    {
        "exactamente 1 fila sobre el ala superior": geo["sobre"].sum(axis=1) == 1,
        "ninguna fila dentro del espesor de un ala": geo["en_ala"].sum(axis=1) == 0,
        "filas = alma + cantonera + 2": geo["n_alma"] + geo["n_cant"] + 2
        == geo["n_filas"],
        "hay filas de cantonera  <=>  Lh > 0": (geo["n_cant"] > 0) == has_Lh,
    },
    index=df.index,
)
print("Filas que incumplen cada comprobación:")
print((~checks).sum().to_string())
inconsistentes = ~checks.all(axis=1)
print(f"\nUniones incoherentes: {int(inconsistentes.sum())} de {len(df)}")
df.loc[
    inconsistentes, ["UNICODE", "e", "d", "Lh", "BEAM_d_mm", "BEAM_tf_mm"] + p_cols[:6]
]
```

```python
df.loc[inconsistentes, ["UNICODE"]]
```

```python
# Las uniones incoherentes (p. ej. Lh > 0 sin filas de cantonera, o una fila dentro del ala) contradicen la
# geometría y confunden al modelo. Pon False para conservarlas y revisarlas a mano.
DROP_INCOHERENTES = True
if DROP_INCOHERENTES and inconsistentes.any():
    print(f"Se eliminan {int(inconsistentes.sum())} uniones incoherentes")
    df = df[~inconsistentes.values].reset_index(drop=True)
    geo = calcular_geometria(df)

df["n_filas_alma"] = geo["n_alma"]  # filas en el alma de la viga
df["n_filas_cantonera"] = geo["n_cant"]  # filas en la cantonera (0 si no hay Lh)
D_total = geo["D"]  # posición de la última fila (mm)

# Posición de cada fila como fracción de D_total (0 = fila inexistente). Vuelta a mm: D = e / row_1_pct.
row_pct = pd.DataFrame(
    np.where(geo["exists"], geo["y"] / D_total[:, None], 0.0),
    index=df.index,
    columns=[f"row_{i}_pct" for i in range(1, 14)],
)
df = pd.concat([df, row_pct], axis=1)
print(df[["n_filas_alma", "n_filas_cantonera"]].value_counts().sort_index().head(12))
```

## HUECO DE TRABAJO DE LOS TORNILLOS
No hay tabla de espacio libre para la herramienta, así que **no se impone una regla externa**: se mide lo que ya cumplen los diseños y se deja que el modelo coloque los tornillos.
Hueco = distancia del eje de la fila de tornillos a la cara de la pieza más próxima (mm); su cociente por el diámetro `db` es adimensional.

| columna | distancia medida |
|---|---|
| `d` (ya existente) | fila 1 ↔ cara exterior del ala superior |
| `hueco_alma_sup` | primera fila del alma ↔ cara interior del ala superior |
| `hueco_alma_inf` | última fila del alma ↔ cara interior del ala inferior |
| `hueco_cant_sup` | primera fila de la cantonera ↔ cara exterior del ala inferior (0 si no hay cantonera) |
| `hueco_alma_lat` | `(g − tw)/2`: eje del tornillo ↔ cara del alma, en horizontal |
| `hueco_min_ratio` | mínimo de los huecos a las alas/alma (incluida `d`) dividido por `db` |
| `paso_min_ratio` | menor separación entre filas consecutivas dividida por `db` |

No se calculan huecos que dependan de `Lh` (último ala de la cantonera): en ~19 % de las uniones con cantonera la última fila no queda donde `Lh` la situaría, así que esa medida no es fiable.
Los umbrales son **solo informativos** (se miden en los datos, no vienen de una norma); solo se descartan las uniones sin hueco físico (`< 0.5·db`).

```python
HUECO_MIN_RATIO = 1.5  # solo para informar: (distancia eje-cara) / db que cumple la inmensa mayoría de los diseños
HUECO_DESCARTAR_BAJO = 0.5  # uniones con algún hueco < 0.5*db no tienen sitio físico para la herramienta: se descartan

db = df["Bolt"].astype(str).str.extract(r"(\d+)")[0].astype(float).values
y = geo["y"]
es_ultima = y >= geo["y_ult"][:, None] - 1e-9
cant = geo["bajo"] & ~es_ultima
tiene_cant = cant.any(axis=1)


def borde(zona, cual):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        ys = np.where(zona, y, np.nan)
        return np.nanmin(ys, axis=1) if cual == "primera" else np.nanmax(ys, axis=1)


df["hueco_alma_sup"] = borde(geo["alma"], "primera") - geo["top_in"]
df["hueco_alma_inf"] = geo["bot_in"] - borde(geo["alma"], "ultima")
df["hueco_cant_sup"] = np.where(
    tiene_cant, borde(cant, "primera") - geo["bot_out"], 0.0
)
df["hueco_alma_lat"] = (df["g"] - df["BEAM_tw_mm"]) / 2.0

nombres = ["d", "hueco_alma_sup", "hueco_alma_inf", "hueco_cant_sup", "hueco_alma_lat"]
gaps = df[nombres].to_numpy(float).copy()
gaps[~tiene_cant, 3] = np.nan  # sin cantonera ese hueco no existe
df["hueco_min_ratio"] = np.nanmin(gaps, axis=1) / db
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    df["paso_min_ratio"] = (
        np.nanmin(np.diff(y, axis=1), axis=1) / db
    )  # separación mínima entre filas consecutivas
quien = pd.Series(
    np.array(nombres)[np.nanargmin(gaps / db[:, None], axis=1)], index=df.index
)

print("Hueco / db por tipo (cuantiles):")
print(
    (pd.DataFrame(gaps / db[:, None], columns=nombres).quantile([0.01, 0.05, 0.5]))
    .round(2)
    .T.to_string()
)
bajo_min = df["hueco_min_ratio"] < HUECO_MIN_RATIO
print(
    f"\nUniones con algún hueco < {HUECO_MIN_RATIO}*db: {int(bajo_min.sum())} ({bajo_min.mean():.1%}) | < 0 (colisión): {int((df['hueco_min_ratio'] < 0).sum())}"
)
print("Dónde está el hueco mínimo en esas uniones:")
print(quien[bajo_min].value_counts().to_string())
print(
    f"\nPaso mínimo entre filas / db: cuantiles 1%, 5%, 50% = {df['paso_min_ratio'].quantile([0.01, 0.05, 0.5]).round(2).tolist()}  (mínimo {df['paso_min_ratio'].min():.2f})"
)
```

```python
descartar = df["hueco_min_ratio"] < HUECO_DESCARTAR_BAJO
print(
    f"Se eliminan {int(descartar.sum())} uniones sin hueco físico para la herramienta (< {HUECO_DESCARTAR_BAJO}*db)"
)
df = df[~descartar].reset_index(drop=True)
df.shape
```

## POSICIÓN DE LAS ALAS, CLASIFICACIÓN DE TORNILLOS Y `Bolt` SIN "M"
Se calcula tras eliminar las uniones sin hueco (se recalcula `geo`, porque ya no coincide con las filas de `df`).

**Distancia de las alas** (borde superior de la chapa → centro del ala, en % de `D`, igual que `row_i_pct`): `ala_sup_pct`, `ala_inf_pct` y `ala_cant_pct` (0 si no hay cantonera).

**Posición de cada tornillo** (`bolt_i_pos`, de arriba a abajo; `ninguno` si la fila no existe):

| etiqueta | significado |
|---|---|
| `superior` / `inferior` | primera fila (sobre el ala superior) / última fila de la chapa |
| `viga_*` | filas del alma de la viga |
| `cantonera_*` | filas de la cantonera |

Sufijos dentro de `viga_` y `cantonera_`: `primero`, `ultimo`, `centro` (zona de 1 fila o central de una zona impar), `centro_sup` / `centro_inf` (las 2 centrales de una zona par), `intermedio_sup` / `intermedio_inf` (el resto).

```python
# Tras el paso anterior se eliminaron uniones: se recalcula la geometría para que sus filas coincidan con df.
geo = calcular_geometria(df)
D_total = geo["D"]
tf = df["BEAM_tf_mm"].values
Lh_mm = df["Lh"].fillna(0).values.astype(float)
tiene_cant = geo["n_cant"] > 0
es_ultima = geo["y"] >= D_total[:, None] - 1e-9
cant = geo["bajo"] & ~es_ultima

# --- 1) Distancia borde superior de la chapa -> centro de cada ala, en % de D (igual que row_i_pct) ---
df["ala_sup_pct"] = (geo["top_out"] + tf / 2) / D_total
df["ala_inf_pct"] = (geo["bot_out"] - tf / 2) / D_total
# SUPUESTO: Lh va de la cara exterior del ala inferior de la viga a la cara exterior del ala de la cantonera,
# y ese ala tiene el espesor del ala de la viga. Si no es así, solo hay que cambiar la línea siguiente.
df["ala_cant_pct"] = np.where(
    tiene_cant, (geo["bot_out"] + Lh_mm - tf / 2) / D_total, 0.0
)

# Comprobación: el centro del ala de la cantonera debe quedar entre la última fila de la cantonera y la última fila.
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    y_ult_cant = np.nanmax(np.where(cant, geo["y"], np.nan), axis=1)
c_mm = df["ala_cant_pct"].values * D_total
ok = (c_mm > y_ult_cant) & (c_mm < D_total)
print(
    f"Ala de cantonera entre su última fila y la final: {ok[tiene_cant].mean():.1%} de {int(tiene_cant.sum())} uniones con cantonera"
)
print(
    df[["ala_sup_pct", "ala_inf_pct", "ala_cant_pct"]].describe().round(3).T.to_string()
)


# --- 2) Clasificación de cada tornillo según su posición ---
def etiquetas_zona(n):
    # etiquetas de las n filas de una zona (viga o cantonera), de arriba a abajo
    if n == 1:
        return ["centro"]
    out = []
    for k in range(n):
        if k == 0:
            out.append("primero")
        elif k == n - 1:
            out.append("ultimo")
        elif n % 2 == 1 and k == n // 2:
            out.append("centro")
        elif n % 2 == 0 and k == n // 2 - 1:
            out.append("centro_sup")
        elif n % 2 == 0 and k == n // 2:
            out.append("centro_inf")
        else:
            out.append("intermedio_sup" if k < n / 2 else "intermedio_inf")
    return out


N_ROWS = 13
pos = np.full((len(df), N_ROWS), "ninguno", dtype=object)
n_alma, n_cant = geo["n_alma"], geo["n_cant"]
for a, c in set(zip(n_alma, n_cant)):
    # orden vertical: 1 fila sobre el ala, filas del alma, filas de la cantonera, fila final
    seq = (
        ["superior"]
        + ["viga_" + l for l in etiquetas_zona(a)]
        + ["cantonera_" + l for l in etiquetas_zona(c)]
        + ["inferior"]
    )
    assert len(seq) == 2 + a + c <= N_ROWS
    pos[(n_alma == a) & (n_cant == c), : len(seq)] = seq

pos_cols = [f"bolt_{i}_pos" for i in range(1, N_ROWS + 1)]
df = pd.concat([df, pd.DataFrame(pos, columns=pos_cols, index=df.index)], axis=1)
print(pd.Series(pos[pos != "ninguno"]).value_counts().to_string())


# --- 3) Bolt sin la "M" (M20 -> 20, diámetro en mm). Si algún valor no tiene número, falla aquí. ---
df["Bolt"] = df["Bolt"].astype(str).str.extract(r"(\d+)")[0].astype(int)
print(df["Bolt"].value_counts().sort_index().to_string())
```

## RATIOS ESTRUCTURALES Y CARGAS

```python
# Ratios adimensionales viga/columna (las dimensiones absolutas BEAM_*_mm / COLUMN_*_mm se descartan en COLUMNAS FINALES).
# NO se añaden ratios con tp, Bp, e o d: son targets y filtrarían la respuesta.
df["r_d_beam_col"] = df["BEAM_d_mm"] / df["COLUMN_d_mm"]
df["r_bf_beam_col"] = df["BEAM_bf_mm"] / df["COLUMN_bf_mm"]
df["r_tw_beam_col"] = df["BEAM_tw_mm"] / df["COLUMN_tw_mm"]
df["r_tf_beam_col"] = df["BEAM_tf_mm"] / df["COLUMN_tf_mm"]

# Cargas: se conservan solo los porcentajes (0-1+); los valores absolutos se descartan.
df["ratio_moment"] = df["Capacity_Moment_Percentage"] / 100.0
df["ratio_shear"] = df["Capacity_Shear_Percentage"] / 100.0
df["ratio_axial"] = df["Capacity_Axial_Percentage"] / 100.0
```

# COLUMNAS FINALES
Se conservan solo las que usa el modelo o sirven para validarlo. Eliminadas y por qué:

| eliminadas | motivo |
|---|---|
| `p1..p12` | sustituidas por `row_i_pct` (+ `e` en mm) |
| `rows_in_top_extension`, `rows_in_bottom_extension`, `num_bolt_rows` | constantes (siempre 1) o redundantes con `n_filas_alma + n_filas_cantonera + 2` |
| `BEAM_*_mm`, `COLUMN_*_mm` | medidas absolutas de perfil: solo sirven para calcular ratios, posiciones y huecos; no entran al modelo |
| `bolt_i_label` (×13) | sustituidas por `bolt_i_pos` (clasificación por zona, ver arriba) |
| `Lp` | largo de la chapa: lo calcula Excel a partir de las filas (era un target encubierto) |
| `DrawingCode`, `Source`, `Normativas`, `National_Annex` | metadatos: no están disponibles al predecir o duplican `standard` |
| `fy/fu_acero`, `fy/fu/fv_tornillo` | función exacta de `grado_acero` y `clase_tornillo` |
| `Capacity_*` (absolutas y %) | sustituidas por `ratio_moment/shear/axial` |
| `AceroEstructural`, `EstandarTornillos` | duplicados de `grado_acero`, `clase_tornillo` |

```python
ID = [
    "UNICODE",
    "BEAM",
    "COLUMN",
]  # identificadores (NO son entradas): split por pares, trazabilidad
ENTRADAS_CAT = ["BEAM_type", "COLUMN_type", "grado_acero", "clase_tornillo"]
# Las medidas absolutas BEAM_*_mm / COLUMN_*_mm se usan arriba para ratios y posiciones; aquí se descartan.
ENTRADAS_NUM = [
    "r_d_beam_col",
    "r_bf_beam_col",
    "r_tw_beam_col",
    "r_tf_beam_col",
    "ratio_moment",
    "ratio_shear",
    "ratio_axial",
]
TARGETS_CLS = ["standard", "Bolt", "SlipCritical", "n_filas_alma", "n_filas_cantonera"]
TARGETS_MM = ["g", "e", "d", "Bp", "tp", "tws", "ts", "tbs", "tst"]
TARGETS_POS = [f"row_{i}_pct" for i in range(1, 14)]
TARGETS_POS_ALAS = [
    "ala_sup_pct",
    "ala_inf_pct",
    "ala_cant_pct",
]  # centro de las alas, en % de D
TARGETS_POS_LABEL = [
    f"bolt_{i}_pos" for i in range(1, 14)
]  # clasificación de cada tornillo
DERIVADAS_Y_AUX = [
    "Hole",
    "Lh",
    "hueco_alma_sup",
    "hueco_alma_inf",
    "hueco_cant_sup",
    "hueco_alma_lat",
    "hueco_min_ratio",
    "paso_min_ratio",
]  # no son entradas; Hole se deriva de Bolt; Lh = largo de la cantonera (0 si no hay)
FINAL = (
    ID
    + ENTRADAS_CAT
    + ENTRADAS_NUM
    + TARGETS_CLS
    + TARGETS_MM
    + TARGETS_POS
    + TARGETS_POS_ALAS
    + TARGETS_POS_LABEL
    + DERIVADAS_Y_AUX
    + ["BEAM_d_mm"]
)
falta = [c for c in FINAL if c not in df.columns]
assert not falta, f"columnas finales ausentes: {falta}"
print("Eliminadas:", [c for c in df.columns if c not in FINAL])
df = df[FINAL].copy()

# Sin imputación silenciosa: si queda algún nulo en las columnas finales es un error de datos.
nulos = df.isna().sum()
assert nulos.sum() == 0, nulos[nulos > 0]
assert (df[["d", "e", "g"]] > 0).all().all()
print(df.shape)
df.head()
```
