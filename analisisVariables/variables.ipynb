```python
# Columnas p1 a p12
p_cols = [f"p{i}" for i in range(1, 13)]

# 1. Calcular la distancia total desde el borde superior hasta el último tornillo
# (e + la suma de todos los pX)
D_total = df["e"] + df[p_cols].sum(axis=1)

# 2. Calcular las posiciones acumuladas en mm
bolt_pos_mm = pd.DataFrame(index=df.index)
bolt_pos_mm["bolt_1_mm"] = df["e"]

# Suma acumulada de las pX
p_cumsum = df[p_cols].cumsum(axis=1)

for i in range(2, 14):  # Tornillos del 2 al 13
  bolt_pos_mm[f"bolt_{i}_mm"] = df["e"] + p_cumsum[f"p{i-1}"]

# 3. Convertir cada posición a porcentaje del total
bolt_pct = pd.DataFrame(index=df.index)
for i in range(1, 14):
  bolt_pct[f"bolt_{i}_pct"] = (bolt_pos_mm[f"bolt_{i}_mm"] / D_total)

# 4. Manejo de tornillos inexistentes (opcional):
# Si una fila tiene menos de 13 filas de tornillos, los tornillos sobrantes
# repetirán el 100%. Si prefieres que valgan 0% para el modelo de IA:
mask_exist = pd.DataFrame(True, index=df.index, columns=bolt_pct.columns)
mask_exist["bolt_1_pct"] = True
for i in range(2, 14):
  # Un tornillo existe si el p correspondiente anterior es mayor que 0
  mask_exist[f"bolt_{i}_pct"] = (df[f"p{i-1}"] > 0) & mask_exist[
      f"bolt_{i-1}_pct"
  ]

# Aplicar máscara: 0 en los tornillos no utilizados
bolt_pct = bolt_pct.where(mask_exist, 0.0)

# Unir las nuevas columnas de porcentaje al DataFrame principal
df_transformed = pd.concat([df, bolt_pct], axis=1)
```

# CLASIFICACIÓN TORNILLOS

```python
def clasificar_tornillos_dataset(row):
  # 1. Calcular las posiciones absolutas de los tornillos activos (en mm)
  e = row["e"]
  p_cols = [f"p{i}" for i in range(1, 13)]

  y_bolts = [e]
  curr_y = e
  for col in p_cols:
    p = row[col]
    if p > 0:
      curr_y += p
      y_bolts.append(curr_y)
    else:
      break

  n_bolts = len(y_bolts)
  # Inicializar lista de 13 etiquetas con 'NO_BOLT'
  labels = ["NO_BOLT"] * 13

  if n_bolts == 0:
    return labels

  # 2. Asignar extremos absolutos
  labels[0] = "GLOBAL_EXT_TOP"
  labels[n_bolts - 1] = "GLOBAL_EXT_BOT"

  # 3. Límites geométricos
  y_beam_top = row["e"] + row["d"] + (row["BEAM_tf_mm"] / 2.0)
  y_beam_bot = (
      row["e"] + row["d"] + row["BEAM_d_mm"] - (row["BEAM_tf_mm"] / 2.0)
  )

  Lh = row["Lh"]
  y_haunch_bot = y_beam_bot + Lh if Lh > 0 else y_beam_bot

  # 4. Asignar índices a cada zona
  indices_beam = []
  indices_haunch = []

  for idx in range(n_bolts):
    y = y_bolts[idx]
    if y_beam_top <= y <= y_beam_bot:
      indices_beam.append(idx)
    elif Lh > 0 and y_beam_bot < y <= y_haunch_bot:
      indices_haunch.append(idx)

  # 5. Función auxiliar para calcular el rol dentro de una zona
  def asignar_roles(indices, prefix):
    N = len(indices)
    for pos, idx in enumerate(indices):
      # Preservar el primer y último tornillo global
      if idx == 0 or idx == (n_bolts - 1):
        continue

      i = pos + 1  # Base 1
      if N == 1:
        role = "SINGLE"
      elif i == 1:
        role = "FIRST"
      elif i == N:
        role = "LAST"
      elif N % 2 != 0:  # Impar
        center = (N + 1) // 2
        if i == center:
          role = "CENTER"
        elif i < center:
          role = "INTERMEDIATE_TOP"
        else:
          role = "INTERMEDIATE_BOTTOM"
      else:  # Par
        center_top = N // 2
        center_bot = center_top + 1
        if i == center_top:
          role = "CENTER_TOP"
        elif i == center_bot:
          role = "CENTER_BOTTOM"
        elif i < center_top:
          role = "INTERMEDIATE_TOP"
        else:
          role = "INTERMEDIATE_BOTTOM"

      labels[idx] = f"{prefix}_{role}"

  asignar_roles(indices_beam, "BEAM")
  asignar_roles(indices_haunch, "HAUNCH")

  return labels
```

```python
label_cols = [f"bolt_{i}_label" for i in range(1, 14)]

# Generar DataFrame con las nuevas etiquetas
df_labels = df.apply(
    clasificar_tornillos_dataset, axis=1, result_type="expand"
)
df_labels.columns = label_cols

# Concatenar al DataFrame original
df = pd.concat([df, df_labels], axis=1)
```

# RATIOS ESTRUCTURALES

```python
def agregar_ratios_estructurales(df):
  # --- 1. Ratios Geométricos Directos ---
  df["r_d_beam_col"] = df["BEAM_d_mm"] / df["COLUMN_d_mm"]
  df["r_bf_beam_col"] = df["BEAM_bf_mm"] / df["COLUMN_bf_mm"]
  df["r_tw_beam_col"] = df["BEAM_tw_mm"] / df["COLUMN_tw_mm"]
  df["r_tf_beam_col"] = df["BEAM_tf_mm"] / df["COLUMN_tf_mm"]

  # Relación Chapa con Viga y Columna
  df["r_tp_tf_beam"] = df["tp"] / df["BEAM_tf_mm"]
  df["r_Bp_bf_col"] = df["Bp"] / df["COLUMN_bf_mm"]
  df["r_Bp_bf_beam"] = df["Bp"] / df["BEAM_bf_mm"]

  # --- 2. Funciones auxiliares para Inercia y Wpl ---
  def calc_section_props(d, bf, tw, tf):
    area = 2 * bf * tf + (d - 2 * tf) * tw
    iy = (tw * (d - 2 * tf) ** 3) / 12.0 + 2 * (
        (bf * (tf**3)) / 12.0 + bf * tf * (((d - tf) / 2.0) ** 2)
    )
    wpl = bf * tf * (d - tf) + (tw * ((d - 2 * tf) ** 2)) / 4.0
    return iy, wpl

  # Inercias y Wpl Viga
  I_beam, Wpl_beam = calc_section_props(
      df["BEAM_d_mm"], df["BEAM_bf_mm"], df["BEAM_tw_mm"], df["BEAM_tf_mm"]
  )

  # Inercias y Wpl Columna
  I_col, Wpl_col = calc_section_props(
      df["COLUMN_d_mm"],
      df["COLUMN_bf_mm"],
      df["COLUMN_tw_mm"],
      df["COLUMN_tf_mm"],
  )

  # Ratios Mecánicos
  df["r_Iy_beam_col"] = I_beam / I_col
  df["r_Wpl_beam_col"] = Wpl_beam / Wpl_col

  # --- 3. Esbelteces Locales ---
  df["slenderness_beam_web"] = (
      df["BEAM_d_mm"] - 2 * df["BEAM_tf_mm"]
  ) / df["BEAM_tw_mm"]
  df["slenderness_col_web"] = (
      df["COLUMN_d_mm"] - 2 * df["COLUMN_tf_mm"]
  ) / df["COLUMN_tw_mm"]
  df["slenderness_col_flange"] = (
      df["COLUMN_bf_mm"] - df["COLUMN_tw_mm"]
  ) / (2 * df["COLUMN_tf_mm"])

  return df


# Aplicar al dataframe
df = agregar_ratios_estructurales(df)
```
