```python
import numpy as np
import pandas as pd


def transformar_topologia_y_cotas(df):
    """Transforma pasos relativos (p_i) en cotas absolutas desde el borde

    superior de la chapa (Y_i = e + sum(p_k)) y clasifica filas por zona
    viga.
    """
    p_cols = [f'p{i}' for i in range(1, 13)]

    def procesar_fila(row):
        pasos = [
            row[col] for col in p_cols if pd.notna(row[col]) and row[col] > 0
        ]
        n_filas = len(pasos) + 1 if pd.notna(row.get('e')) else 0

        res = {
            'n_sup': 0,
            'n_alma': 0,
            'n_cantonera': 0,
            'n_inf': 0,
            'num_filas': n_filas,
        }

        # Inicializamos 12 cotas Y1..Y12 (correspondientes al tornillo final de cada paso p1..p12)
        for i in range(1, 13):
            res[f'Y{i}'] = 0.0

        if n_filas < 2:
            return pd.Series(res)

        # 1. Cotas Absolutas (Origen Y=0 en Borde Superior de la Chapa)
        e = row['e']
        Y_coords = [e]  # Fila 1 está en cota 'e'

        # Y1 = e + p1, Y2 = e + p1 + p2...
        y_acum = e
        for p in pasos:
            y_acum += p
            Y_coords.append(y_acum)

        # Guardar en Y1..Y12 (Y_coords tiene len = n_filas)
        for idx, y_val in enumerate(Y_coords[1:], start=1):
            res[f'Y{idx}'] = y_val

        # 2. Fronteras físicas de la viga
        d = row.get('d', 0)
        h_beam = row['BEAM_d_mm']

        y_ala_sup = e + d
        y_ala_inf = e + d + h_beam

        # 3. Clasificación topológica por zonas
        n_debajo = 0
        for y_val in Y_coords:
            if y_val < y_ala_sup:
                res['n_sup'] += 1  # Tornillos por encima de la viga
            elif y_val <= y_ala_inf:
                res['n_alma'] += 1  # Tornillos entre las alas
            else:
                n_debajo += 1  # Tornillos por debajo de la viga

        # 4. Desglose cantonera / fuera
        if n_debajo > 0:
            res['n_inf'] = 1
            res['n_cantonera'] = n_debajo - 1
        else:
            res['n_inf'] = 0
            res['n_cantonera'] = 0

        return pd.Series(res)

    df_transformado = df.apply(procesar_fila, axis=1)
    return pd.concat([df, df_transformado], axis=1)


# ==============================================================================
# FUNCIÓN INVERSA PARA RECUPERAR LOS PASOS p_i DE TALLER
# ==============================================================================
def convertir_cotas_a_pasos(df_prediccion):
    """Convierte cotas absolutas Y1..Y12 y borde 'e' de vuelta a pasos relativos

    p1..p12.
    """
    df_res = df_prediccion.copy()

    # p1 = Y1 - e
    if 'Y1' in df_res.columns and 'e' in df_res.columns:
        mask_y1 = df_res['Y1'] > 0
        df_res.loc[mask_y1, 'p1'] = (
            df_res.loc[mask_y1, 'Y1'] - df_res.loc[mask_y1, 'e']
        )
        df_res.loc[~mask_y1, 'p1'] = 0.0

    # p_i = Y_i - Y_{i-1} para i >= 2
    for i in range(2, 13):
        col_actual = f'Y{i}'
        col_anterior = f'Y{i-1}'
        col_paso = f'p{i}'

        if col_actual in df_res.columns and col_anterior in df_res.columns:
            mask_existe = df_res[col_actual] > 0
            df_res.loc[mask_existe, col_paso] = (
                df_res.loc[mask_existe, col_actual]
                - df_res.loc[mask_existe, col_anterior]
            )
            df_res.loc[~mask_existe, col_paso] = 0.0

    return df_res

df_resultado = transformar_topologia_y_cotas(df_final3)
```

# QUITAR LOS DISEÑOS QUE DIFIEREN EN CANTONERA

```python
def eliminar_duplicados_con_cantonera(df: pd.DataFrame, load_bin: float = 5.0) -> pd.DataFrame:
    """Elimina diseños con cantonera cuando existe una alternativa equivalente sin ella."""
    if load_bin <= 0:
        raise ValueError("load_bin debe ser mayor que cero")

    df_work = df.copy()
    group_parts = {}

    for column in ["BEAM", "COLUMN", "AceroEstructural", "EstandarTornillos", "standard"]:
        if column in df_work.columns:
            group_parts[column] = df_work[column].astype("string").fillna("<NA>")

    for column in [
        "Capacity_Moment_Percentage",
        "Capacity_Shear_Percentage",
        "Capacity_Axial_Percentage",
    ]:
        if column in df_work.columns:
            values = pd.to_numeric(df_work[column], errors="coerce")
            group_parts[column] = (
                np.floor(values / load_bin)
                .astype("Int64")
                .astype("string")
                .fillna("<NA>")
            )

    if group_parts:
        group_key = pd.concat(group_parts, axis=1)
    else:
        group_key = pd.Series(0, index=df_work.index, name="_group_key")

    def positive_column(column: str) -> pd.Series:
        if column not in df_work.columns:
            return pd.Series(False, index=df_work.index)
        return pd.to_numeric(df_work[column], errors="coerce").fillna(0).gt(0)

    has_cantonera = positive_column("n_cantonera") | positive_column("Lh")

    if group_parts:
        has_option_without_cantonera = (
            (~has_cantonera)
            .groupby([group_key[column] for column in group_key.columns], dropna=False, sort=False)
            .transform("any")
        )
    else:
        has_option_without_cantonera = pd.Series(
            (~has_cantonera).any(), index=df_work.index
        )

    keep = (~has_cantonera) | ~has_option_without_cantonera
    df_filtered = df_work.loc[keep].copy()

    print("[+] Limpieza completada:")
    print(f"    - Filas iniciales : {len(df_work)}")
    print(f"    - Filas eliminadas: {len(df_work) - len(df_filtered)} (diseños duplicados con cantonera)")
    print(f"    - Filas finales   : {len(df_filtered)}")

    return df_filtered

eliminar_duplicados_con_cantonera(df_final3)
```
