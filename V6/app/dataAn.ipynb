```python
import pandas as pd


def etiquetas_zona(nombre_zona, cantidad):
    """
    Devuelve las etiquetas de los tornillos interiores de una zona,
    ordenados de arriba hacia abajo.
    """
    cantidad = int(cantidad)

    if cantidad <= 0:
        return []

    if cantidad == 1:
        return [f"{nombre_zona} central"]

    if cantidad == 2:
        return [
            f"{nombre_zona} primero",
            f"{nombre_zona} inferior"
        ]

    return (
        [f"{nombre_zona} primero"]
        + [f"{nombre_zona} central"] * (cantidad - 2)
        + [f"{nombre_zona} inferior"]
    )


def clasificar_fila(fila):
    n_cantonera = int(fila["n_cantonera"])
    n_alma = int(fila["n_alma"])

    clasificaciones = ["superior"]

    clasificaciones += etiquetas_zona(
        nombre_zona="viga",
        cantidad=n_alma
    )

    clasificaciones += etiquetas_zona(
        nombre_zona="cantonera",
        cantidad=n_cantonera
    )

    clasificaciones.append("inferior")

    return pd.Series(
        clasificaciones,
        index=[
            f"tornillo_{i}"
            for i in range(1, len(clasificaciones) + 1)
        ]
    )
```

```python
# Generar las columnas de clasificación
clasificacion_tornillos = df.apply(
    clasificar_fila,
    axis=1
)

# Ordenar tornillo_1, tornillo_2, ..., tornillo_10...
columnas_tornillos = sorted(
    clasificacion_tornillos.columns,
    key=lambda columna: int(columna.split("_")[-1])
)

clasificacion_tornillos = clasificacion_tornillos[columnas_tornillos]

# Añadirlas al final del DataFrame original
df = pd.concat(
    [df, clasificacion_tornillos],
    axis=1
)
```

```python
def añadir_transiciones_p(df, max_p=10):
    """
    Para cada separación p_i crea una transición:

        p1_transition = tornillo_1 -> tornillo_2
        p2_transition = tornillo_2 -> tornillo_3
        ...

    Las separaciones inexistentes reciben UNUSED.
    """
    resultado = df.copy()

    for i in range(1, max_p + 1):
        col_p = f"p{i}"
        col_superior = f"tornillo_{i}"
        col_inferior = f"tornillo_{i + 1}"
        col_transition = f"p{i}_transition"

        existe_p = (
            pd.to_numeric(
                resultado[col_p],
                errors="coerce",
            )
            .fillna(0.0)
            .gt(0.0)
        )

        tornillo_superior = (
            resultado[col_superior]
            .fillna("no_usado")
            .astype(str)
        )

        tornillo_inferior = (
            resultado[col_inferior]
            .fillna("no_usado")
            .astype(str)
        )

        resultado[col_transition] = (
            tornillo_superior
            + "__"
            + tornillo_inferior
        )

        resultado.loc[
            ~existe_p,
            col_transition,
        ] = "UNUSED"

    return resultado
```

```python
df = añadir_transiciones_p(
    df,
    max_p=10
)
```

```python
def validar_clasificacion_tornillos(df, max_p=10):
    resultados = []

    for indice, fila in df.iterrows():
        n_p_esperadas = (
            int(fila["n_alma"])
            + int(fila["n_cantonera"])
            + 1
        )

        p_activas = sum(
            pd.notna(fila[f"p{i}"])
            and float(fila[f"p{i}"]) > 0
            for i in range(1, max_p + 1)
        )

        transiciones_activas = sum(
            fila[f"p{i}_transition"] != "UNUSED"
            for i in range(1, max_p + 1)
        )

        resultados.append({
            "indice": indice,
            "p_esperadas": n_p_esperadas,
            "p_activas": p_activas,
            "transiciones_activas": transiciones_activas,
            "conteo_p_correcto": p_activas == n_p_esperadas,
            "conteo_transiciones_correcto": (
                transiciones_activas == n_p_esperadas
            ),
        })

    validacion = pd.DataFrame(resultados).set_index("indice")

    validacion["valido"] = (
        validacion["conteo_p_correcto"]
        & validacion["conteo_transiciones_correcto"]
    )

    return validacion
```

```python
def crear_dataset_separaciones(df, max_p=10):
    registros = []

    for i in range(1, max_p + 1):
        col_p = f"p{i}"
        col_transition = f"p{i}_transition"

        columnas = [
            col_p,
            col_transition,
            "Bolt",
            "BEAM_d_mm",
            "n_alma",
            "n_cantonera",
        ]

        parte = df[columnas].copy()

        parte = parte.rename(
            columns={
                col_p: "separacion_mm",
                col_transition: "transicion",
            }
        )

        parte["numero_p"] = i

        registros.append(parte)

    separaciones = pd.concat(
        registros,
        ignore_index=True,
    )

    separaciones["separacion_mm"] = pd.to_numeric(
        separaciones["separacion_mm"],
        errors="coerce",
    ).fillna(0.0)

    separaciones["Bolt"] = (
        separaciones["Bolt"]
        .astype(str)
        .str.replace("M", "", regex=False)
    )

    separaciones["Bolt"] = pd.to_numeric(
        separaciones["Bolt"],
        errors="coerce",
    )

    separaciones = separaciones.loc[
        separaciones["separacion_mm"].gt(0.0)
        & separaciones["transicion"].ne("UNUSED")
    ].reset_index(drop=True)

    return separaciones
```

```python
separaciones = crear_dataset_separaciones(df)

resumen_separaciones = (
    separaciones
    .groupby(
        ["transicion", "Bolt"],
        dropna=False,
    )["separacion_mm"]
    .agg(
        numero_casos="size",
        minimo="min",
        percentil_05=lambda x: x.quantile(0.05),
        percentil_10=lambda x: x.quantile(0.10),
        mediana="median",
        media="mean",
        maximo="max",
    )
    .reset_index()
    .sort_values(
        ["transicion", "Bolt"]
    )
)
```

```python
resumen_separaciones["grupo_suficiente"] = (
    resumen_separaciones["numero_casos"] >= 20
)
```
