```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path().cwd() / "data"
# 1. Columnas del Parquet
df_parquet = pd.read_parquet(DATA_PATH / "datos_modelo.parquet")
print("PARQUET COLUMNS:", df_parquet.columns.tolist())
print("PARQUET SAMPLE:\n", df_parquet.head(1).to_dict(orient='records'))

# 2. Columnas del CSV de perfiles
df_perfiles_eu = pd.read_csv(DATA_PATH / "tekla_eu.csv")
print("\nCOLUMNAS PERFILES EUROPEOS:", df_perfiles_eu.columns.tolist())

df_perfiles_cad = pd.read_csv(DATA_PATH / "tekla_cad.csv")
print("\nCOLUMNAS PERFILES CANADIENSE:", df_perfiles_cad.columns.tolist())

df_perfiles_uk = pd.read_csv(DATA_PATH / "tekla_uk.csv")
print("\nCOLUMNAS PERFILES BRITANICOS:", df_perfiles_uk.columns.tolist())

df_perfiles_faltante = pd.read_csv(DATA_PATH / "tekla_faltante.csv")
print("\nCOLUMNAS PERFILES OTROS:", df_perfiles_faltante.columns.tolist())
```
