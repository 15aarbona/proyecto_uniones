```python
excel = pd.read_excel(
    Path().cwd().parent / "data" / "CONEXIONES_MOMENTO_ATORNILLADAS_CHAPA_FRONTAL.xlsx"
)
excel.head()
```

```python
excel.columns
```

```python
cols_interes = [
    "Unicode",
    "Capacity_Moment_Percentage",
    "Capacity_Shear_Percentage",
    "Capacity_Axial_Percentage",
]
df_excel_sub = excel[cols_interes]

# 3. Cruzar ambas tablas mediante la columna 'Unicode'
df = pd.merge(
    data, df_excel_sub, on="Unicode", how="left"
)
```

```python
df.head()
```

```python
df["ratio_moment"] = df["Capacity_Moment_Percentage"] / 100
df["ratio_shear"] = df["Capacity_Shear_Percentage"] / 100
df["ratio_axial"] = df["Capacity_Axial_Percentage"] / 100
```

```python
df.drop(columns=["Capacity_Moment", "Capacity_Shear", "Capacity_Axial", "Capacity_Moment_Percentage", "Capacity_Shear_Percentage", "Capacity_Axial_Percentage"], inplace=True)
```
