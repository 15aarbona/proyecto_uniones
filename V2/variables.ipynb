## DISTANCIAS

```python
import numpy as np
import pandas as pd


def calcular_distancias_union(df):
    p_cols = [f"p{i}" for i in range(1, 13)]

    def procesar_fila(row):
        # 1. Filtrar los pasos p1..p12 que existen en esta unión (no NaN y > 0)
        pasos = [row[col] for col in p_cols if pd.notna(row[col]) and row[col] > 0]
        n_filas = len(pasos) + 1  # Número de filas de tornillos

        if n_filas < 2:
            return pd.Series(
                {
                    "E-S": np.nan,
                    "I-C": np.nan,
                    "E-C": np.nan,
                    "tiene_cantonera": False,
                    "num_filas_tornillos": n_filas,
                }
            )

        # 2. Coordenadas Y absolutas de cada fila de tornillos (origen Y=0 en el borde superior)
        e = row["e"]
        d = row["d"]
        h_beam = row["BEAM_d_mm"]

        Y = [e]
        for p in pasos:
            Y.append(Y[-1] + p)

        # 3. Ubicación del ala inferior de la viga principal
        Y_ala_inf_viga = e + d + h_beam

        # 4. Extremo Superior (E-S): Paso entre la fila 1 y fila 2 (p1)
        ES = pasos[0]

        # 5. Intermedio Viga (I-C): Paso que abarca el ala inferior de la viga
        idx_IC = None
        for k in range(1, len(Y)):
            if Y[k - 1] <= Y_ala_inf_viga <= Y[k]:
                idx_IC = k - 1
                break

        IC = pasos[idx_IC] if idx_IC is not None else np.nan

        # 6. Extremo Cantonera (E-C): Fila del ala de la cantonera
        # Si hay filas de tornillos después del ala inferior de la viga, existe cantonera
        tiene_cantonera = False
        EC = np.nan

        if idx_IC is not None and (idx_IC + 1) < len(pasos):
            tiene_cantonera = True
            EC = pasos[-1]  # El último paso corresponde al extremo de la cantonera

        # 7. Pasos intermedios del alma D_i (entre el ala sup y el ala inf de la viga)
        fin_intermedios = idx_IC if idx_IC is not None else len(pasos)
        D_intermedios = pasos[1:fin_intermedios]

        # Resultado base para la fila
        res = {
            "E-S": ES,
            "I-C": IC,
            "E-C": EC,
            "tiene_cantonera": tiene_cantonera,
            "num_filas_tornillos": n_filas,
        }

        # Generar dinámicamente columnas D_1, D_2, etc.
        for idx, d_val in enumerate(D_intermedios, start=1):
            res[f"D_{idx}"] = d_val

        return pd.Series(res)

    # Aplicar la función a todo el DataFrame y concatenar los resultados
    distancias_df = df.apply(procesar_fila, axis=1)
    return pd.concat([df, distancias_df], axis=1)


# Ejemplo de uso con tu dataframe:
df_resultado = calcular_distancias_union(df_final3)
```
